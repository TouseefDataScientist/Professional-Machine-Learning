```python
from azure.core.credentials import AzureKeyCredential
from azure.ai.textanalytics import TextAnalyticsClient

# 1. Your Azure Details (Endpoint and Key)
endpoint = "https://mylanguageserviceresource.cognitiveservices.azure.com/"
key = "Your api "

# Authenticating the client
def authenticate_client():
    ta_credential = AzureKeyCredential(key)
    text_analytics_client = TextAnalyticsClient(
            endpoint=endpoint, 
            credential=ta_credential)
    return text_analytics_client

def main():
    client = authenticate_client()
    
    # 2. Input text to analyze
    documents = [
        "My name is ammar I am from Rawalpindi i completed my fsc from 502 College after that i currently enrolled in microsoft certification which are conducted in tevta under the cherif minister skill development program."
    ]
    
    # Sentiment Analysis (Positive, Negative, or Neutral)
    print("--- Running Sentiment Analysis ---")
    sentiment_result = client.analyze_sentiment(documents=documents)[0]
    print(f"Overall Sentiment: {sentiment_result.sentiment.upper()}")
    print(f"Scores: Positive={sentiment_result.confidence_scores.positive:.2f}; Neutral={sentiment_result.confidence_scores.neutral:.2f}; Negative={sentiment_result.confidence_scores.negative:.2f}\n")
    
    # Key Phrases Extraction (Extracting important words/topics)
    print("--- Extracting Key Phrases ---")
    key_phrase_result = client.extract_key_phrases(documents=documents)[0]
    if not key_phrase_result.is_error:
        print("Key Phrases:")
        for phrase in key_phrase_result.key_phrases:
            print(f" - {phrase}")
    else:
        print(f"Error: {key_phrase_result.error}")

if __name__ == "__main__":
    main()
```

```python
from azure.core.credentials import AzureKeyCredential
from azure.ai.textanalytics import TextAnalyticsClient
from pypdf import PdfReader  # PDF read karne ke liye library

# 1. Your Azure Details (Endpoint and Key)
endpoint = "https://mylanguageserviceresource.cognitiveservices.azure.com/"

# Apni active Azure API key yahan dalein:
key = "your api" 

# Authenticating the client
def authenticate_client():
    ta_credential = AzureKeyCredential(key)
    text_analytics_client = TextAnalyticsClient(
            endpoint=endpoint, 
            credential=ta_credential)
    return text_analytics_client

# PDF file se text extract karne ka function
def extract_text_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    full_text = ""
    
    # Har page ka text extract karke combine karna
    for page in reader.pages:
        text = page.extract_text()
        if text:
            full_text += text + "\n"
        
    return full_text.strip()

def main():
    client = authenticate_client()
    
    # 2. Apni PDF Resume file ka path yahan dein
    # Agar file same folder mein ha toh sirf naam likhein (e.g., "resume.pdf")
    pdf_file_path = "C:/Users/Ammar Javed/Downloads/Hamza Mehmood CV.pdf"
    
    try:
        print("--- Extracting Text from PDF ---")
        resume_text = extract_text_from_pdf(pdf_file_path)
        
        if not resume_text:
            print("Error: PDF se koi text nahi mila ya file khali ha.")
            return
            
        # Azure expects a list of documents
        documents = [resume_text]
        
        # Sentiment Analysis (Positive, Negative, or Neutral)
        print("--- Running Sentiment Analysis on Resume ---")
        sentiment_result = client.analyze_sentiment(documents=documents)[0]
        print(f"Overall Sentiment: {sentiment_result.sentiment.upper()}")
        print(f"Scores: Positive={sentiment_result.confidence_scores.positive:.2f}; Neutral={sentiment_result.confidence_scores.neutral:.2f}; Negative={sentiment_result.confidence_scores.negative:.2f}\n")
        
        # Key Phrases Extraction (Extracting important words/topics)
        print("--- Extracting Key Phrases ---")
        key_phrase_result = client.extract_key_phrases(documents=documents)[0]
        if not key_phrase_result.is_error:
            print("Key Phrases:")
            for phrase in key_phrase_result.key_phrases:
                print(f" - {phrase}")
        else:
            print(f"Error: {key_phrase_result.error}")
            
    except FileNotFoundError:
        print(f"Error: Di gayi file '{pdf_file_path}' nahi mili. Baraye meherbani sahi path check karein.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Execute the main function directly for the notebook environment
main()
```
